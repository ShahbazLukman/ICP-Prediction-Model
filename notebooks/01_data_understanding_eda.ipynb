This notebook is intentionally beginner-friendly. Run cells from top to bottom.

**Business question:** Can lead behavior help predict whether a course enquiry converts within 30 days?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'raw' / 'synthetic_course_leads.csv')
df.head()
```

```python
print('Shape:', df.shape)
df.info()
```

```python
df.isna().sum().sort_values(ascending=False).head(10)
```

```python
df['purchased_within_30_days'].value_counts(normalize=True)
```

```python
sns.countplot(data=df, x='purchased_within_30_days')
plt.title('Purchase Distribution')
plt.show()
```

```python
purchase_by_course = df.groupby('course_enquiry', dropna=False)['purchased_within_30_days'].mean().sort_values(ascending=False)
print(purchase_by_course)
purchase_by_course.plot(kind='bar')
plt.ylabel('Purchase rate')
plt.title('Purchase Rate by Enquired Course')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
behavior_cols = ['website_visits','course_page_views','pricing_page_views','whatsapp_responses','calls_answered','followup_count','days_since_last_activity']
df.groupby('purchased_within_30_days')[behavior_cols].mean().T
```
